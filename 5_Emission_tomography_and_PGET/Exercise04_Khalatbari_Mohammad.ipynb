## Task 1 (Direct problem)
Recall the groundwater filtration example from the lecture slides: for
any permittivity $\sigma \in C^1[0,1]$ and source $g \in C[0,1]$, let $u \in C^2[0,1]$ be the
solution of
\begin{align}
-\frac{\partial}{\partial x} \left( \sigma \frac{\partial u}{\partial x} \right) (x) &= g(x) \text{ on } [0,1] \\
\frac{\partial u}{\partial x}(0) &= 0 \nonumber \\
u(1) &= 0 \nonumber
\end{align}
Discretize the equation for the variable $U = [u_0,...,u_N]^T \in \mathbb{R}^{N+1}$, where $u_0 = u_1$ and
$u_N = 0$ according to the boundary values and $\sigma \geq \sigma_0$ for some constant
$\sigma_0 > 0$.

Given some functions $g$ and $\sigma$, write a Python routine `solveU(sigma)` for the solution operator $F_g : \sigma \mapsto U$ of
the direct problem using finite difference approximations of the derivative. The following script is a good starting point.

### Task 1 solution

Set $h=1/N$ and $x_i=ih$ for $i=0,\ldots,N$. Store the $N$ permittivity values at the interval midpoints $x_{j+1/2}=(j+1/2)h$. At each midpoint, approximate $\sigma u'$ by the flux $q_{j+1/2}=\sigma_{j+1/2}(u_{j+1}-u_j)/h$. Then at each interior node,

$$
-\frac{q_{i+1/2}-q_{i-1/2}}{h}=g(x_i),
\qquad i=1,\ldots,N-1,\qquad g(x_i)=\sin(\pi x_i).
$$

The matrices $L_1$ and $L_2$ take these two differences, each divided by $h$. The interior block is $-L_2\operatorname{diag}(\sigma)L_1$. Appending the rows for $u_1-u_0=0$ and $u_N=0$ gives $N+1$ equations for $N+1$ nodal values. Positive $\sigma$ makes the system nonsingular. With zero right-hand side, the flux is constant and the left boundary makes it zero, so all nodal values are equal and the right boundary sets them to zero.

The clean solve uses both boundary conditions. The noisy curve adds independent Gaussian measurement noise with standard deviation $0.01$ to every sample. See §4.1, p. 19, Eq. (26) in the [lecture notes](../../Lecture_notes_v1.2.pdf).


```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as op

# set dimension and a lower limit to sigma
N = 101
sigma0 = 0.2

# contruct ground truth sigma using an exponential mollifier 
sigma_true = sigma0*np.ones((N,))
with np.errstate(divide='ignore'): # Silence division by zero errors
    sigma_true[(N//3):2*(N//3)] = np.exp(-1/(1 - np.linspace(-1,1, N//3, endpoint=True)**2)) + sigma0
xs = (np.arange(N) + 0.5) / N  # Permittivity is sampled at interval midpoints.

# plot the ground truth sigma
plt.plot(xs, sigma_true)
plt.title("True sigma")
plt.xlabel("x")
plt.ylabel(r"$\sigma(x)$")
plt.show()
```

```python
def solveU(sigma):
    """Return the nodal solution for the given midpoint permittivity values."""
    N = len(sigma)
    h = 1 / N
    x_interior = h * np.arange(1, N)
    g = np.sin(np.pi * x_interior).reshape((-1, 1))

    # Difference adjacent nodal values, then adjacent midpoint fluxes.
    L1 = np.diff(np.eye(N + 1), axis=0) / h
    L2 = np.diff(np.eye(N), axis=0) / h

    # Boundary condition: u_1 - u_0 = 0
    b1 = np.zeros((1, N + 1))
    b1[0, :2] = [-1, 1]

    # Boundary condition: u_N = 0
    b2 = np.zeros((1, N + 1))
    b2[0, -1] = 1

    # Multiplying each row of L1 by sigma gives the midpoint fluxes.
    sigma = sigma.reshape((-1, 1))
    K = -L2 @ (sigma * L1)
    K = np.concatenate((K, b1, b2), axis=0)

    extra = np.zeros((2, 1))
    g0 = np.concatenate((g, extra), axis=0)

    u, *_ = np.linalg.lstsq(K, g0, rcond=None)
    return u.reshape((-1, 1))


# Solve the forward problem and add reproducible measurement noise.
u = solveU(sigma_true)
x_nodes = np.linspace(0, 1, u.size)
plt.plot(x_nodes, u, label="Clean data")

delta = 0.01
rng = np.random.default_rng(42)
un = u + rng.normal(0, delta, size=u.shape)

plt.plot(x_nodes, un, label="Noisy data")
plt.xlabel("x")
plt.ylabel(r"$u(x)$")
plt.legend()
plt.show()
```
