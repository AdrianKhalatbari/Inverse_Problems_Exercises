## Task 1 (Direct problem)
Recall the groundwater filtration example from the lecture slides: for
any permittivity $\sigma \in C^1[0,1]$ and source $g \in C[0,1]$, let $u \in C^2[0,1]$ be the
solution of
\begin{align}
-\frac{\partial}{\partial x} \left( \sigma \frac{\partial u}{\partial x} \right) (x) &= g(x) \text{ on } [0,1] \\
\frac{\partial u}{\partial x}(0) &= 0 \nonumber \\
u(1) &= 0 \nonumber
\end{align}
Discretize the equation for the variable $U = [u_0,...,u_N]^T \in \mathbb{R}^{N+1}$, where $u_0 = u_1$ and
$u_N = 0$ according to the boundary values and $\sigma \geq \sigma_0$ for some constant
$\sigma_0 > 0$.

Given some functions $g$ and $\sigma$, write a Python routine `solveU(sigma)` for the solution operator $F_g : \sigma \mapsto U$ of
the direct problem using finite difference approximations of the derivative. The following script is a good starting point.

### Task 1 solution

Set $h=1/N$ and $x_i=ih$ for $i=0,\ldots,N$. Store the $N$ permittivity values at the interval midpoints $x_{j+1/2}=(j+1/2)h$. At each midpoint, approximate $\sigma u'$ by the flux $q_{j+1/2}=\sigma_{j+1/2}(u_{j+1}-u_j)/h$. Then at each interior node,

$$
-\frac{q_{i+1/2}-q_{i-1/2}}{h}=g(x_i),
\qquad i=1,\ldots,N-1,\qquad g(x_i)=\sin(\pi x_i).
$$

The matrices $L_1$ and $L_2$ take these two differences, each divided by $h$. The interior block is $-L_2\operatorname{diag}(\sigma)L_1$. Appending the rows for $u_1-u_0=0$ and $u_N=0$ gives $N+1$ equations for $N+1$ nodal values. Positive $\sigma$ makes the system nonsingular. With zero right-hand side, the flux is constant and the left boundary makes it zero, so all nodal values are equal and the right boundary sets them to zero.

The clean solve uses both boundary conditions. The noisy curve adds independent Gaussian measurement noise with standard deviation $0.01$ to every sample. See §4.1, p. 19, Eq. (26) in the [lecture notes](../../Lecture_notes_v1.2.pdf).


```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as op

# set dimension and a lower limit to sigma
N = 101
sigma0 = 0.2

# contruct ground truth sigma using an exponential mollifier 
sigma_true = sigma0*np.ones((N,))
with np.errstate(divide='ignore'): # Silence division by zero errors
    sigma_true[(N//3):2*(N//3)] = np.exp(-1/(1 - np.linspace(-1,1, N//3, endpoint=True)**2)) + sigma0
xs = (np.arange(N) + 0.5) / N  # Permittivity is sampled at interval midpoints.

# plot the ground truth sigma
plt.plot(xs, sigma_true)
plt.title("True sigma")
plt.xlabel("x")
plt.ylabel(r"$\sigma(x)$")
plt.show()
```

```python
def solveU(sigma):
    """Return the nodal solution for the given midpoint permittivity values."""
    N = len(sigma)
    h = 1 / N
    x_interior = h * np.arange(1, N)
    g = np.sin(np.pi * x_interior).reshape((-1, 1))

    # Difference adjacent nodal values, then adjacent midpoint fluxes.
    L1 = np.diff(np.eye(N + 1), axis=0) / h
    L2 = np.diff(np.eye(N), axis=0) / h

    # Boundary condition: u_1 - u_0 = 0
    b1 = np.zeros((1, N + 1))
    b1[0, :2] = [-1, 1]

    # Boundary condition: u_N = 0
    b2 = np.zeros((1, N + 1))
    b2[0, -1] = 1

    # Multiplying each row of L1 by sigma gives the midpoint fluxes.
    sigma = sigma.reshape((-1, 1))
    K = -L2 @ (sigma * L1)
    K = np.concatenate((K, b1, b2), axis=0)

    extra = np.zeros((2, 1))
    g0 = np.concatenate((g, extra), axis=0)

    u, *_ = np.linalg.lstsq(K, g0, rcond=None)
    return u.reshape((-1, 1))


# Solve the forward problem and add reproducible measurement noise.
u = solveU(sigma_true)
x_nodes = np.linspace(0, 1, u.size)
plt.plot(x_nodes, u, label="Clean data")

delta = 0.01
rng = np.random.default_rng(42)
un = u + rng.normal(0, delta, size=u.shape)

plt.plot(x_nodes, un, label="Noisy data")
plt.xlabel("x")
plt.ylabel(r"$u(x)$")
plt.legend()
plt.show()
```

## Task 2 (Tikhonov regularization)
Using the routine produced in previous exercise, reconstruct the Tikhonov
regularized solution
\begin{equation*}
\sigma^\delta_\alpha := \argmin_{\sigma} \lbrace \| F_g(\sigma) − y^\delta \|^2 + \alpha \|\sigma − \sigma_0 \|^2 \rbrace
\end{equation*}
for a given regularization parameter $\alpha$ using gradient descent algorithm [`fmin_cg`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.fmin_cg.html) (conjugate gradient) from `scipy.optimize`. Parameters and data are provided in the following Python script.

### Task 2 solution

I use the noisy Task 1 data as $y^\delta$, with $\alpha=0.05$ and $\sigma_0=0.2$ in the penalty. Both runs use the Task 1 forward solve and nonlinear conjugate gradients with SciPy's finite-difference gradient.

The supplied run reaches its 200-iteration cap with negative coefficients. To keep coefficients positive, I rerun CG in variables $z$ with $\sigma_i=e^{z_i}$, starting at the same $\sigma_i=0.01$ and allowing up to 400 iterations. This minimizes the same objective over strictly positive coefficients. In the displayed Task 2 objective, $0.2$ is the prior level. The supplied start is already below it.

The default finite-difference step produced a precision-loss warning in log coordinates. A step of $10^{-6}$ reduces the effect of subtracting nearly equal objective values.

```python
from numpy.linalg import norm

alpha = 0.05
x0 = np.full(N, 0.01)

def objFun(sigma):
    """Squared data misfit plus the Tikhonov penalty."""
    residual = solveU(sigma) - un
    return norm(residual)**2 + alpha * norm(sigma - sigma0)**2

status_message = ("converged", "iteration limit reached", "precision loss", "NaN encountered")

# Run the supplied unconstrained problem in sigma coordinates.
sigma_raw, value_raw, _, _, flag_raw = op.fmin_cg(
    objFun, x0, maxiter=200, full_output=True, disp=False
)
print(f"Initial objective: {objFun(x0):.6g}")
print(f"Direct CG: {status_message[flag_raw]}, objective {value_raw:.6g}, "
      f"minimum sigma {sigma_raw.min():.6f}")

def positive_objective(log_sigma):
    """Evaluate the same objective using positive permittivity."""
    return objFun(np.exp(log_sigma))

# A larger finite-difference step reduces cancellation in the computed objective.
log_recon, value, _, _, flag, iterates = op.fmin_cg(
    positive_objective, np.log(x0), epsilon=1e-6, maxiter=400, full_output=True, retall=True, disp=False
)
recon_cg = np.exp(log_recon)
u_recon = solveU(recon_cg)

print(f"Log-parameter CG: {status_message[flag]} after {len(iterates) - 1} iterations, "
      f"objective {value:.6g}")
print(f"Permittivity range: [{recon_cg.min():.6f}, {recon_cg.max():.6f}]")
print(f"Data residual norm: {norm(u_recon - un):.6f}")

fig, (ax_sigma, ax_data) = plt.subplots(1, 2, figsize=(10, 3.6), layout="constrained")
ax_sigma.plot(xs, sigma_true, label="True")
ax_sigma.plot(xs, recon_cg, label="Reconstructed")
ax_sigma.set(xlabel="x", ylabel=r"$\sigma(x)$", title="Permittivity reconstruction")
ax_sigma.legend(loc="upper left")

ax_data.plot(x_nodes, un, label="Noisy data", alpha=0.7)
ax_data.plot(x_nodes, u_recon, label="Fitted data")
ax_data.set(xlabel="x", ylabel=r"$u(x)$", title="Fit to the measurements")
ax_data.legend()
plt.show()
```

CG met its gradient stopping criterion in log coordinates. The fitted data have an RMS residual of 0.0132. The recovered peak is lower and broader than the true peak. The penalty pulls the coefficients toward the prior, and the forward map depends on weighted sums of reciprocal permittivities. Some positive coefficients fall below $0.2$, which is used as a reference level in this reconstruction.
